## Demonstração do Controller de Contas Bancárias

Este notebook demonstra o funcionamento do controller de contas bancárias da API, cobrindo as seguintes operações:
- Criar contas bancária
- Buscar todas as contas bancárias
- Buscar contas bancária por ID
- Atualizar contas bancária
- Desativar, Restaurar e Deletar do Banco de dados a contas bancária


from fastapi.testclient import TestClient
from fastapi import FastAPI

from accounts.controller import accounts_router

app = FastAPI()
app.include_router(accounts_router)

client = TestClient(app)

## Criar conta bancária de teste


user_payload = {
   "name": "Conta Bancária de Alan",
   "type": "Poupança",
   "account_number": "AM20131125254632633269",
   "balance": 50000,
}

response = client.post("/accounts/create", json=user_payload)
print("Status:", response.status_code)
print("Resposta:", response.json())
created_account = response.json()
account_id = created_account["id"]

## Buscar todos as contas bancárias ativos

response = client.get("/accounts/all")
print("Status:", response.status_code)
for account in response.json():
    print(account)

## Buscar conta bancária por ID

response = client.get(f"/accounts/{account_id}")
print("Status:", response.status_code)
print("Conta:", response.json())

## Atualizar dados da conta bancária

update_payload = {
    "balance": 100000.0
}

response = client.patch(f"/accounts/{account_id}/update", json=update_payload)
print("Status:", response.status_code)
print("Atualizado:", response.json())

## Desativar conta bancária

response = client.delete(f"/accounts/{account_id}/delete")
print("Status:", response.status_code)
print("Conta desativada:", response.json())

## Restaurar conta deletada
-- Opção de restaurar do banco de dados a conta (Precisa estar previamente Desativada)

response = client.post(f"/accounts/{account_id}/restore")
print("Status:", response.status_code)
print("Restaurado:", response.json())

## Deletar do Banco de Dados a conta deletado
-- Opção de deletar do banco de dados a conta (Precisa estar previamente Desativada)


```python
response = client.delete(f"/accounts/{account_id}/force-delete")
print("Status:", response.status_code)
print("Forçando a Deleção no banco de dados:", response.json())
```
