## Demonstração do Controller de Contas Bancárias

Este notebook demonstra o funcionamento do controller de contas bancárias da API, cobrindo as seguintes operações:
- Criar conta bancária (`POST /accounts/create`)
- Buscar todas as contas bancárias ativas (`GET /accounts/all`)
- Buscar conta bancária por ID (`GET /accounts/{id}`)
- Atualizar conta bancária (`PATCH /accounts/{id}/update`)
- Desativar conta bancária (`DELETE /accounts/{id}/delete`)
- Restaurar conta desativada (`POST /accounts/{id}/restore`)
- Deletar permanentemente do banco (`DELETE /accounts/{id}/force-delete`)


## Setup do Teste com FastAPI e TestClient

Import e configuração do FastAPI com o router `accounts`.

```python
from fastapi.testclient import TestClient
from fastapi import FastAPI

from accounts.controller import accounts_router

app = FastAPI()
app.include_router(accounts_router)

client = TestClient(app)
```

## Criar Conta Bancária de Teste

**Endpoint:** `POST /accounts/create`  
**Descrição:** Cria uma nova conta bancária com os dados fornecidos no payload.

```python
user_payload = {
   "name": "Conta Bancária de Alan",
   "type": "Poupança",
   "account_number": "AM20131125254632633269",
   "balance": 50000,
}

response = client.post("/accounts/create", json=user_payload)
print("Status:", response.status_code)
print("Resposta:", response.json())
created_account = response.json()
account_id = created_account["id"]
```

## Buscar Todas as Contas Bancárias Ativas

**Endpoint:** `GET /accounts/all`  
**Descrição:** Retorna todas as contas bancárias que estão ativas no sistema.

```python
response = client.get("/accounts/all")
print("Status:", response.status_code)
for account in response.json():
    print(account)
```

## Buscar Conta Bancária por ID

**Endpoint:** `GET /accounts/{id}`  
**Descrição:** Retorna os dados de uma conta bancária específica, identificada pelo ID.

```python
response = client.get(f"/accounts/{account_id}")
print("Status:", response.status_code)
print("Conta:", response.json())
```

## Atualizar Dados da Conta Bancária

**Endpoint:** `PATCH /accounts/{id}/update`  
**Descrição:** Atualiza os campos fornecidos da conta bancária (ex: saldo, tipo).

```python
update_payload = {
    "balance": 100000.0
}

response = client.patch(f"/accounts/{account_id}/update", json=update_payload)
print("Status:", response.status_code)
print("Atualizado:", response.json())
```

## Desativar Conta Bancária

**Endpoint:** `DELETE /accounts/{id}/delete`  
**Descrição:** Marca a conta como desativada (soft delete), sem removê-la do banco.

```python
response = client.delete(f"/accounts/{account_id}/delete")
print("Status:", response.status_code)
print("Conta desativada:", response.json())
```

## Restaurar Conta Bancária Desativada

**Endpoint:** `POST /accounts/{id}/restore`  
**Descrição:** Restaura uma conta que foi previamente desativada.

```python
response = client.post(f"/accounts/{account_id}/restore")
print("Status:", response.status_code)
print("Restaurado:", response.json())
```

## Deletar Conta Bancária Permanentemente

**Endpoint:** `DELETE /accounts/{id}/force-delete`  
**Descrição:** Deleta permanentemente a conta do banco de dados (hard delete). A conta precisa estar desativada antes.


```python
response = client.delete(f"/accounts/{account_id}/force-delete")
print("Status:", response.status_code)
print("Forçando a Deleção no banco de dados:", response.json())
```
