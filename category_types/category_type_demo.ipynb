## Demonstração do Controller de Tipo de Categoria

Este notebook demonstra o funcionamento do controller de tipo de categorias da API, cobrindo as seguintes operações:
- Criar tipo de categoria (`POST /category_types/create`)
- Buscar todos os tipo de categorias ativos (`GET /category_types/all`)
- Atualizar tipo de categoria (`PATCH /category_types/{id}/update`)
- Desativar tipo de categoria (`DELETE /category_types/{id}/delete`)


## Setup do Teste com FastAPI e TestClient

Import e configuração do FastAPI com o router `category_types`.

```python
from fastapi.testclient import TestClient
from fastapi import FastAPI

from category_types.controller import category_types_router

app = FastAPI()
app.include_router(category_types_router)

client = TestClient(app)
```

## Criar Tipo de Categoria de Teste

**Endpoint:** `POST /category_types/create`  
**Descrição:** Cria um novo tipo de categoria com os dados fornecidos no payload.

```python
user_payload = {
    "name": "Gastos",
    "is_positive": False,
}

response = client.post("/category_types/create", json=user_payload)
print("Status:", response.status_code)
print("Resposta:", response.json())
created_category_type = response.json()
category_type_id = created_category_type["id"]
```

## Buscar Todos os Tipo de Categorias Ativos

**Endpoint:** `GET /category_types/all`  
**Descrição:** Retorna todos os tipo de categorias que estão ativos no sistema.

```python
response = client.get("/category_types/all")
print("Status:", response.status_code)
for user in response.json():
    print(user)
```

## Atualizar Dados do Tipo de Categoria

**Endpoint:** `PATCH /category_types/{id}/update`  
**Descrição:** Atualiza os campos fornecidos do tipo de categoria (ex: nome da categoria).

```python
update_payload = {
    "name": "Despesas"
}

response = client.patch(f"/category_types/{category_type_id}/update", json=update_payload)
print("Status:", response.status_code)
print("Atualizado:", response.json())
```

## Deletar Tipo de Categoria Permanentemente

**Endpoint:** `DELETE /category_types/{id}/delete`  
**Descrição:** Deleta permanentemente o tipo de categoria do banco de dados (hard delete). 

```python
response = client.delete(f"/category_types/{category_type_id}/delete")
print("Status:", response.status_code)
print("Forçando a Deleção no banco de dados:", response.json())
```
