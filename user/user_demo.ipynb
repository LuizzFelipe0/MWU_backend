## Demonstração do Controller de Usuário

Este notebook demonstra o funcionamento do controller de usuários da API, cobrindo as seguintes operações:
- Criar usuário
- Buscar todos os usuários
- Buscar usuário por ID
- Atualizar usuário
- Deletar e restaurar usuário


```python
from fastapi.testclient import TestClient
from fastapi import FastAPI

from user.controller import users_router

app = FastAPI()
app.include_router(users_router)

client = TestClient(app)
```

## 👤 Criar usuários de teste

Vamos criar alguns usuários com dados fictícios para testar as funcionalidades.

```python
user_payload = {
    "first_name": "Alan",
    "last_name": "Gamer",
    "cpf": "961.456.040-18",
    "email": "agamer@example.com",
    "password": "Strong@Password1986",
    "manual_balance": 500.0
}

response = client.post("/users/create", json=user_payload)
print("Status:", response.status_code)
print("Resposta:", response.json())
created_user = response.json()
user_id = created_user["id"]
```

## 📋 Buscar todos os usuários ativos

```python
response = client.get("/users/all")
print("Status:", response.status_code)
for user in response.json():
    print(user)
```

## 🔍 Buscar usuário por ID

```python
response = client.get(f"/users/{user_id}")
print("Status:", response.status_code)
print("Usuário:", response.json())
```

## 📝 Atualizar dados do usuário

```python
update_payload = {
    "manual_balance": 1000.0
}

response = client.patch(f"/users/{user_id}/update", json=update_payload)
print("Status:", response.status_code)
print("Atualizado:", response.json())
```

## ❌ Deletar usuário

```python
response = client.delete(f"/users/{user_id}/delete")
print("Status:", response.status_code)
print("Usuário deletado:", response.json())
```

## 🧟 Restaurar usuário deletado

```python
response = client.post(f"/users{user_id}/restore")
print("Status:", response.status_code)
print("Restaurado:", response.json())
```
